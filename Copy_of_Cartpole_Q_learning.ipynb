```python

import pickle
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML, display

# Load trained model
with open("cartpole_qtable.pkl", "rb") as f:
    saved = pickle.load(f)

q_table = saved["q_table"]
bins = saved["bins"]


def discretize_state(state):
    return tuple(
        int(np.digitize(state[i], bins[i]))
        for i in range(4)
    )


# Create environment with image rendering
env = gym.make("CartPole-v1", render_mode="rgb_array")

observation, info = env.reset()
state = discretize_state(observation)

frames = []
total_reward = 0
terminated = False
truncated = False

# Run the trained agent
while not (terminated or truncated):

    # Choose the best learned action
    action = int(np.argmax(q_table[state]))

    observation, reward, terminated, truncated, info = (
        env.step(action)
    )

    state = discretize_state(observation)
    total_reward += reward

    # Capture the simulation frame
    frames.append(env.render())

env.close()

# Animate the frames
fig, ax = plt.subplots(figsize=(7, 5))
ax.axis("off")
image = ax.imshow(frames[0])

def update(frame_number):
    image.set_data(frames[frame_number])
    return [image]

ani = animation.FuncAnimation(
    fig,
    update,
    frames=len(frames),
    interval=30,
    blit=True
)

plt.close(fig)
display(HTML(ani.to_jshtml()))

print("Evaluation reward:", total_reward)
print("Steps survived:", len(frames))
```
